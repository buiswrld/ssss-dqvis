```python
from datasets import load_dataset
import pandas as pd
```

```python
dataset = load_dataset(
    "HIDIVE/DQVis",
    split="train",
    streaming=True
)
```

```python
subset = list(
    dataset.shuffle(seed=42, buffer_size=10000).take(5000)
)
```

```python
df = pd.DataFrame(subset)
```

```python
print("Number of rows:", len(df))
print("Columns:", df.columns.tolist())
```

```python
print("Total rows:", len(df))
print("Unique query_base values:", df["query_base"].nunique())
print("Duplicate query_base rows:", df["query_base"].duplicated().sum())
```

```python
grouped = df.groupby("query_base").agg(
    number_of_paraphrases=("query", "count"),
    paraphrases=("query", list)
).reset_index()

grouped = grouped[grouped["number_of_paraphrases"] > 1]

display(grouped.head(10))
```
